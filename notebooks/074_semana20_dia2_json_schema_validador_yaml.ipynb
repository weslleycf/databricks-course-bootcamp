## 📖 Teoria — YAML de fluxo + JSON Schema

Cada fluxo é definido por um YAML:

```yaml
fluxo: metas
nome: Metas de Vendas
campos:
  - nome: vendedor
    tipo: string
    obrigatorio: true
  - nome: meta
    tipo: numero
    min: 0
  - nome: mes
    tipo: mes
```

O **JSON Schema** valida a ESTRUTURA do YAML (campos obrigatórios, tipos permitidos) antes de processar — evita config quebrada.


### 💻 Na prática — JSON Schema do YAML

Defina o schema que todo YAML de fluxo deve respeitar.


````python
# JSON Schema do YAML de fluxo
json_schema = {
  "type": "object",
  "required": ["fluxo", "nome", "campos"],
  "properties": {
    "fluxo": {"type": "string"},
    "nome": {"type": "string"},
    "campos": {
      "type": "array",
      "items": {
        "type": "object",
        "required": ["nome", "tipo"],
        "properties": {
          "nome": {"type": "string"},
          "tipo": {"enum": ["string", "numero", "data", "mes", "bool"]},
          "obrigatorio": {"type": "boolean"},
          "min": {"type": "number"},
          "max": {"type": "number"}
        }
      }
    }
  }
}
print(json.dumps(json_schema, indent=2))
````

````python
# Validar o YAML contra o schema (jsonschema)
import yaml, json
from jsonschema import validate, ValidationError
yaml_fluxo = """
fluxo: metas
nome: Metas de Vendas
campos:
  - nome: vendedor
    tipo: string
    obrigatorio: true
"""
dados = yaml.safe_load(yaml_fluxo)
try:
    validate(instance=dados, schema=json_schema)
    print("YAML válido!")
except ValidationError as e:
    print("YAML inválido:", e.message)
````

### 💻 Na prática — Teste com YAML inválido

Um YAML sem o campo obrigatório deve falhar.


````python
# YAML inválido (sem nome)
yaml_ruim = "fluxo: metas\n"  # falta nome e campos
try:
    validate(instance=yaml.safe_load(yaml_ruim), schema=json_schema)
    print("ERRO: deveria falhar")
except ValidationError as e:
    print("Corretamente bloqueado:", e.message)
````

> 🎯 **Dica de prova**: Portfólio: JSON Schema valida a configuração ANTES de processar — mesma filosofia de expectations no DLT: falha cedo, falha claro.
