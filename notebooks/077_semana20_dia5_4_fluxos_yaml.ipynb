## 📖 Teoria — Os 4 fluxos do varejo

| Fluxo | Campos | Regras |
|---|---|---|
| **Metas** | vendedor, meta, mês | meta ≥ 0, mês válido |
| **Descontos** | produto, percentual, validade | 0–100%, data futura |
| **Preços** | produto, preço_novo, vigência | preço > 0 |
| **Clientes** | nome, email, cidade | email válido |


### 💻 Na prática — Criando os fluxos

Salve cada fluxo como YAML versionado.


```python
# Fluxo: Metas
metas_yaml = """
fluxo: metas
nome: Metas de Vendas
campos:
  - nome: vendedor
    tipo: string
    obrigatorio: true
  - nome: meta
    tipo: numero
    min: 0
  - nome: mes
    tipo: mes
"""
salvar_versao("metas", metas_yaml, "admin")
print("Fluxo metas salvo.")
```

```python
# Fluxo: Descontos
descontos_yaml = """
fluxo: descontos
nome: Descontos Promocionais
campos:
  - nome: produto
    tipo: string
    obrigatorio: true
  - nome: percentual
    tipo: numero
    min: 0
    max: 100
  - nome: validade
    tipo: data
"""
salvar_versao("descontos", descontos_yaml, "admin")
print("Fluxo descontos salvo.")
```

```python
# Fluxos: Preços e Clientes
precos_yaml = """
fluxo: precos
nome: Atualização de Preços
campos:
  - nome: produto
    tipo: string
    obrigatorio: true
  - nome: preco_novo
    tipo: numero
    min: 0.01
"""
clientes_yaml = """
fluxo: clientes
nome: Cadastro de Clientes
campos:
  - nome: nome
    tipo: string
    obrigatorio: true
  - nome: email
    tipo: string
  - nome: cidade
    tipo: string
"""
salvar_versao("precos", precos_yaml, "admin")
salvar_versao("clientes", clientes_yaml, "admin")
print("4 fluxos salvos e versionados.")
```

### 💻 Na prática — Validando todos

Rode o JSON Schema em cada fluxo salvo.


```python
# Validar os 4 fluxos
for fid in ["metas", "descontos", "precos", "clientes"]:
    y = spark.sql(f"SELECT yaml_def FROM workspace.app.fluxo_versoes WHERE fluxo_id='{fid}' ORDER BY versao DESC LIMIT 1").collect()[0][0]
    try:
        validate(instance=yaml.safe_load(y), schema=json_schema)
        print(f"{fid}: válido")
    except Exception as e:
        print(f"{fid}: INVÁLIDO -> {e.message}")
```

> 🎯 **Dica de prova**: Portfólio: 4 fluxos reais com regras variadas provam o motor — cada tipo de regra (min/max, enum, data) é um caso de teste.
