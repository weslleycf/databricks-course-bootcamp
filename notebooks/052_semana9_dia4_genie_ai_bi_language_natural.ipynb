## 📖 Teoria — O que é o Genie

O **Genie** é o assistente conversacional de BI do Databricks: você pergunta em linguagem natural ('qual o país com mais receita em novembro?') e ele gera/executa SQL sobre as tabelas que você autorizou.

- **Genie spaces**: coleções de tabelas + instruções para o assistente
- **Genie Code**: modo que mostra o SQL gerado (transparência e correção)
- **Ontologia** (DAIS 2026, 🔑): define métricas e cálculos padrão — o Genie não erra conta porque conhece a definição oficial


### 💻 Na prática — Criando um Genie space

1. **AI/BI → Genie → New space**.
2. Adicione as tabelas: `workspace.ouro.vendas_por_dia`, `workspace.ouro.receita_por_pais`, `workspace.ouro.top_produtos`.
3. Escreva instruções (prompt do space): 'Sempre use receita como Quantity*UnitPrice; responda em português'.
4. Pergunte: 'Qual o top 5 de países por receita?'


### 💻 Na prática — Genie Code

No Genie, ative o **Code mode** para ver o SQL gerado — aprenda com ele e valide a resposta.


```python
# O que o Genie gera (exemplo do SQL por trás da resposta)
sql_exemplo = """
SELECT Country, SUM(Quantity * UnitPrice) AS receita
FROM workspace.bronze.vendas_bronze
GROUP BY Country
ORDER BY receita DESC
LIMIT 5
"""
print(sql_exemplo)
print("Valide: rode este SQL você mesmo e compare com a resposta do Genie.")
```
