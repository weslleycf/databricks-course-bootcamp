## 📖 Teoria — O que são Repos

**Databricks Repos** integra o seu repositório Git (GitHub/GitLab/Bitbucket) ao workspace: os notebooks viram arquivos versionados, editáveis no branch certo, com PR e merge pelo Git. É o padrão para times.

Sem Repos, você tem notebooks soltos no workspace — sem versionamento real.


### 💻 Na prática — Conectando o Repo

1. **Workspace → Add → Repo**.
2. URL do seu repositório GitHub (o criado na Semana 2).
3. Autentique (Personal Access Token com escopo `repo`).
4. O repositório aparece no workspace como uma pasta com `workspace`.
5. Navegue: os `.ipynb` deste curso podem ser importados para dentro do repo.


### 💻 Na prática — Branch workflow no Repos

Trabalhe sempre em branch e integre via PR — o padrão de produção.


```python
# Via UI do Repos
print("""
1. No repo, botão branch > criar branch: feature/semana6
2. Edite notebooks nessa branch
3. Commit + Push (botões no topo do repo)
4. No GitHub: crie o PR e faça merge
5. No Repos: pull para atualizar main
""")
print("Repos sincroniza com Git — commits e pushes direto do workspace.")
```

## 📖 Teoria — Workspace files vs notebooks

**Workspace files** são arquivos de código puros (`.py`, `.yml`) versionáveis — o DLT e os DABs usam workspace files. Notebooks têm formato `.ipynb` e também ficam no Git via Repos.
