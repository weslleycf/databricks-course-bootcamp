## 📖 Teoria — O que é tracing

O **MLflow Tracing** registra **spans** (etapas) de uma chamada: retrieval, montagem do prompt, chamada do LLM, resposta. Permite **debugar** onde o RAG errou (contexto ruim? prompt errado? LLM?).

### 💻 Na prática — Analisando o trace

1. **Experiments → run mais recente → Traces**.
2. Veja os spans: retriever (chunks + scores), prompt final, LLM (tokens).
3. Identifique: o contexto recuperado era relevante? O prompt estava certo?


```python
# Ver traces via API
traces = mlflow.search_traces(experiment_names=[mlflow.get_experiment().name])
for t in traces:
    print("Trace:", t.info.trace_id)
    for span in t.data.spans:
        print(f"  span: {span.name} | inputs: {str(span.inputs)[:80]}")
```
