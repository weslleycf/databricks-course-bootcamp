## 📖 Teoria — Os 7 pilares de GenAI em produção

1. **Qualidade mensurável** (avaliação contínua)
2. **Custo controlado** (gateway, cache, modelo certo)
3. **Latência aceitável** (endpoint, otimização de contexto)
4. **Segurança e privacidade** (PII, guardrails, RLS)
5. **Governança** (versões, auditoria, permissões)
6. **Observabilidade** (traces, métricas, alertas)
7. **Escala** (serving, auto-scaling, quota)


## 📖 Teoria — Mosaic AI Model Serving

O **Mosaic AI Model Serving** (2026; ex-Databricks Model Serving) serve: FMA (hospedados) e **modelos custom** (seus MLmodels/LangChain) via endpoint REST com auto-scaling.

Na Free Edition: endpoints de FMA com quota; **sem GPU/provisioned throughput/custom** — custom é 🔑 trial.


### 💻 Na prática — Servindo FMA

Crie um endpoint de FMA (ou use o já existente) e consulte via REST.


```python
# Consultar endpoint de FMA via REST (o padrão de produção)
import requests, json
host = spark.conf.get("spark.databricks.workspaceUrl")
token = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()
resp = requests.post(
    f"https://{host}/serving-endpoints/databricks-llama-3-1-70b/invocations",
    headers={"Authorization": f"Bearer {token}",
             "Content-Type": "application/json"},
    json={"messages": [{"role": "user", "content": "Resuma o Lakehouse em 1 frase."}],
          "temperature": 0.2})
print(resp.json()["choices"][0]["message"]["content"])
```
