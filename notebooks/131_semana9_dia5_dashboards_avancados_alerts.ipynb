## 📖 Teoria — BI profissional no Databricks SQL

- **Dashboards**: múltiplas visualizações com filtros (parâmetros) linkados
- **Alerts**: condições com notificação (e-mail/Slack)
- **Result caching**: o SQL Warehouse guarda resultados por até 24h — consultas iguais retornam instantaneamente (e sem DBU!)
- **Schedules**: queries agendadas (ex.: 8h diária)


### 💻 Na prática — Dashboard com filtros

Crie um dashboard com parâmetro de país que filtra todas as visualizações.


```python
%sql
-- Query com parâmetro {{pais}}
SELECT Country, DATE_TRUNC("month", InvoiceDate) AS mes,
       SUM(Quantity * UnitPrice) AS receita
FROM workspace.bronze.vendas_bronze
WHERE Country = {{pais}}
GROUP BY 1, 2
ORDER BY mes
```

### 💻 Na prática — Alertas e agendamento

1. Salve a query `alerta_receita_diaria` (receita de hoje vs ontem).
2. Crie o **alert**: se variação < -20% → notificar.
3. Agende a query para rodar 07:50 (antes do expediente).


```python
# Query de alerta (receita do dia vs dia anterior)
sql_alerta = """
WITH rec AS (
  SELECT DATE_TRUNC('day', InvoiceDate) dia, SUM(Quantity*UnitPrice) receita
  FROM workspace.bronze.vendas_bronze GROUP BY 1)
SELECT (receita - LAG(receita) OVER (ORDER BY dia)) / LAG(receita) OVER (ORDER BY dia) * 100 AS var_pct
FROM rec ORDER BY dia DESC LIMIT 1
"""
print(sql_alerta)
```

> 🎯 **Dica de prova**: Result caching: consultas repetidas retornam do cache (rápido, sem compute). O cache é invalidado quando a tabela muda. Pergunta típica de DEP: 'por que a 2a execução é mais rápida?' → result caching.
