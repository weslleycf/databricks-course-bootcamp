## 📖 Teoria — Versionar tudo

Em GenAI, o 'modelo' inclui: o LLM base + **prompt** + **chunks/retriever** + dados de avaliação. Mudou o prompt? É uma nova versão.

Boas práticas:
- Prompt como **código** (Git) + registrado como artifact no MLflow
- Golden set versionado (tabela Delta)
- Modelo/prompt/avaliação no mesmo run → rastreabilidade completa


### 💻 Na prática — Versionando o prompt

Registre o prompt do RAG como artifact de uma run.


```python
# Prompt versionado no MLflow
import mlflow
with mlflow.start_run(run_name="prompt_v3"):
    mlflow.log_text(system_prompt, "system_prompt_v3.txt")
    mlflow.log_param("versao_prompt", "v3")
    mlflow.log_param("chunk_size", 400)
    mlflow.log_param("retriever", "hibrido_rrf")
    print("Prompt e configuração versionados na run.")
```

```python
# Golden set versionado como tabela
import pandas as pd
golden = spark.createDataFrame(pd.DataFrame({
    "id": [1, 2, 3],
    "pergunta": ["P1", "P2", "P3"],
    "resposta_esperada": ["R1", "R2", "R3"],
    "versao": ["v3"]
}))
golden.write.mode("overwrite").saveAsTable("workspace.prata.golden_set")
print("Golden set versionado: workspace.prata.golden_set")
```

### 💻 Na prática — Avaliação contínua

Agende a avaliação do RAG: job diário roda golden set e registra métricas no MLflow — queda de métrica dispara alerta.


```python
# Job de avaliação contínua (conceito)
print("""
1. Notebook de avaliação (Semana 12.1) vira job diário
2. Lê o golden set da tabela
3. Roda o RAG e mlflow.evaluate
4. Compara com o baseline registrado
5. Alerta se faithfulness caiu > 5%
""")
print("Monitoramento contínuo = GenAI em produção de verdade.")
```
