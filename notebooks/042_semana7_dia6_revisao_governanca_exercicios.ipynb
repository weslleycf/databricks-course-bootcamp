## 📖 Teoria — Mapa mental da governança

```
Unity Catalog
 ├─ 3 níveis (catalog.schema.object)
 ├─ managed (DROP apaga) vs external (DROP preserva)
 ├─ Volumes (arquivos)
 ├─ Permissões (GRANT/REVOKE por nível)
 ├─ RLS (linhas) + Column Masking (colunas)
 ├─ Dynamic views (current_user)
 ├─ Linhagem + tags + auditoria (system tables)
 ├─ Federation (ler externo) + Sharing (compartilhar)
 └─ LGPD/GDPR (pseudonimização, TTL, apagamento)
```

**6.** Para compartilhar uma tabela com um parceiro externo:
