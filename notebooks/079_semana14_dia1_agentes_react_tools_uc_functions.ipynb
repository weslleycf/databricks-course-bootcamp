## 📖 Teoria — Agente vs LLM

Um **LLM** responde texto. Um **agente** é um LLM + **ferramentas** + **loop**: ele raciocina, decide chamar uma ferramenta, observa o resultado e decide de novo.

**Ciclo ReAct** (Reason + Act):
```
pergunta → raciocínio → ação (tool) → observação → raciocínio... → resposta
```


## 📖 Teoria — Unity Catalog Functions como ferramentas

O UC pode guardar **funções SQL/Python** que o agente chama como ferramentas — com governança (permissões, auditoria, linhagem). É o jeito nativo do Databricks de dar tools ao agente.

`CREATE FUNCTION ... LANGUAGE PYTHON` (ou SQL) → visível como tool.


### 💻 Na prática — Criando UC functions

Crie funções úteis para o agente de vendas.


````python
%sql
-- Função SQL: receita por período
CREATE OR REPLACE FUNCTION workspace.prata.receita_periodo(de STRING, ate STRING)
RETURNS DOUBLE
RETURN (SELECT SUM(receita_total) FROM workspace.ouro.vendas_por_dia
        WHERE data_venda BETWEEN de AND ate);
SELECT workspace.prata.receita_periodo('2024-11-01', '2024-11-30') AS receita_novembro;
````

````python
# Função Python no UC
spark.sql("""
CREATE OR REPLACE FUNCTION workspace.prata.ticket_medio()
RETURNS DOUBLE
LANGUAGE PYTHON
AS $$
    return 42.5
$$
""")
print("UC functions criadas: receita_periodo e ticket_medio.")
````

### 💻 Na prática — Chamando a função via SQL

O agente chamará essas funções; aqui você valida que funcionam.


````python
%sql
SELECT workspace.prata.receita_periodo('2024-01-01', '2024-12-31') AS receita_2024
````
