## 📖 Teoria — Databricks CLI

O CLI (`databricks`) permite tudo o que a UI faz, via terminal: jobs, clusters, pipelines, bundles, tokens, secrets. É a base de automação e CI/CD.

**REST API 2.0**: os mesmos recursos via HTTP (jobs, catalogs, warehouses...). CLI é um cliente da API.


### 💻 Na prática — Comandos essenciais

Rode localmente (ou em %sh no notebook, se o domínio estiver liberado).


```python
# Terminal local
!databricks auth login
!databricks jobs list
!databricks pipelines list
!databricks catalogs list
print("CLI conectado: jobs, pipelines, catálogos listados.")
```

## 📖 Teoria — Secrets — nunca credenciais em código

Credenciais (senhas, tokens, chaves de API) **nunca** devem ficar em notebooks ou código. Use **Secret Scopes**: cofre gerenciado (Databricks-backed) ou integrado (Vault/Key Vault em produção).

Acesso no notebook: `dbutils.secrets.get(scope='meu_scope', key='senha')`.


### 💻 Na prática — Criando e usando secrets

No terminal (local) com o CLI conectado ao workspace.


```python
# Terminal local
!databricks secrets create-scope curso
!databricks secrets put-secret curso senha_api --string-value "abc123"
print("Scope curso criado com a secret senha_api.")
```

```python
# Uso no notebook — NUNCA imprima o valor em produção
senha = dbutils.secrets.get(scope="curso", key="senha_api")
print("Secret lida com sucesso (tamanho:", len(senha), "caracteres)")
```

> 🎯 **Dica de prova**: A DEP cobra: secrets scopes (Databricks-backed vs Key Vault), dbutils.secrets.get, e a regra de nunca expor credenciais. Na Free, scopes avançados (Key Vault) exigem conta paga.
