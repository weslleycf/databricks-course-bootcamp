## 📖 Teoria — O UC como camada única de governança

O Unity Catalog centraliza: catálogos, schemas, tabelas, views, volumes, funções, modelos ML, permissões, linhagem e auditoria. Tudo em um namespace de 3 níveis — a resposta do Databricks ao 'quem tem acesso ao quê'.


## 📖 Teoria — Managed vs External

**Managed table**: o UC gerencia o storage e o ciclo de vida. `DROP TABLE` apaga os dados. Criada com `CREATE TABLE` (sem LOCATION).

**External table**: aponta para um storage **fora do UC** via **external location** (credencial de nuvem). `DROP TABLE` não apaga os arquivos.

**Storage credential**: credencial de nuvem (IAM/service principal) que autoriza o Databricks a acessar seu bucket. **External location** = credencial + caminho (ex.: `s3://meu-bucket/lake`).

> ⚠️ A Free Edition **não tem** external locations — estude o conceito e valide no trial pago (Semana 7, dia com 🔑).


## 📖 Teoria — Volumes

**Volumes** são diretórios governados do UC para arquivos (não tabelas): modelos, parquets de staging, imagens, dados brutos. Caminho: `/Volumes/catalog/schema/volume/...`. Na Free Edition funcionam normalmente.


### 💻 Na prática — Managed + Volumes na prática

Crie tabelas gerenciadas e um volume; veja a diferença de ciclo de vida.


```python
%sql
-- Managed table
CREATE SCHEMA IF NOT EXISTS workspace.prata;
CREATE OR REPLACE TABLE workspace.prata.teste_managed (id INT, nome STRING) USING DELTA;
INSERT INTO workspace.prata.teste_managed VALUES (1, 'a'), (2, 'b');
SELECT * FROM workspace.prata.teste_managed;
```

```python
%sql
-- Volume gerenciado
CREATE VOLUME IF NOT EXISTS workspace.prata.vol_modelos;
SHOW VOLUMES IN workspace.prata;
```

```python
# Acessar o volume via caminho
path = "/Volumes/workspace/prata/vol_modelos"
spark.createDataFrame([("m1", 0.95)], ["modelo", "acuracia"]) \
    .write.mode("overwrite").parquet(f"{path}/resultado.parquet")
display(spark.read.parquet(f"{path}/resultado.parquet"))
```

> 🎯 **Dica de prova**: DEA 2026: UC vale ~30%! Memorize: managed = UC gerencia + DROP apaga; external = LOCATION + DROP preserva; Volumes = arquivos governados; external location = credencial + caminho.
