## 📖 Teoria — mlflow.evaluate

O `mlflow.evaluate` roda **avaliações padronizadas** e registra no MLflow:

- Modelos clássicos: RMSE, MAE, R2, precisão/recall, curvas
- LLMs (LLM-as-judge): faithfulness, answer relevance, ... (Semana 12)
- Modelos de texto: perplexity, toxicity

Vantagem: métricas comparáveis entre runs e auditáveis.

> 🎯 **Dica de prova**: MLP/GenAI: `mlflow.evaluate` aceita `model_type` (regressor/classifier/llm) e `baseline_model`. É o padrão para comparar modelos.
