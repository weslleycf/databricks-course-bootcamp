## 📖 Teoria — Escala bilionária

Para 1B+ vetores:
- **Particionar** por metadata (região, tenant) — busca em partições relevantes
- **Quantização** (reduzir dimensões/precisão) para caber em memória
- **HNSW por partição** + **filtros de metadata antes** da busca
- **Rerank só no top-N** (cross-encoder é caro)

Regra: filtre primeiro, busque depois, rerank no final.


### 💻 Na prática — Arquitetura de retrieval

Desenhe a arquitetura do seu retrieval em escala.


```python
# Decisões de arquitetura
arquitetura = """
1. Particionar vetores por pais/categoria (metadata)
2. Índice HNSW por partição
3. Busca: filtro de metadata -> k-NN na partição
4. Rerank: cross-encoder no top-20 -> top-3
5. Cache semântico (gateway) para perguntas repetidas
"""
print(arquitetura)
```

```python
# Rerank nativo (🔑 trial)
print("""
1. Mosaic AI Reranker (ou Lakebase rerank)
2. POST {query, documents: [top-20]}
3. Retorna reordenados por relevância
4. Use o top-3 como contexto
""")
print("Rerank nativo: melhora Recall@5 sem mudar o índice.")
```

### 💻 Na prática — Custo da busca em escala

Equilibre latência × custo: HNSW com dimensão menor, cache e rerank seletivo.
