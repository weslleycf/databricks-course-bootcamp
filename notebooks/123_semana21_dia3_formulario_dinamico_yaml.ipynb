## 📖 Teoria — Formulário dirigido por configuração

O YAML define os campos → o app **gera o formulário** (text, number, date, select). Novo fluxo = novo formulário **sem código** — a mágica da parametrização por YAML.


### 💻 Na prática — Gerador de formulário

Renderize os campos do YAML como widgets Streamlit.


```python
# Gerador de formulário a partir do YAML
import streamlit as st
def renderizar_formulario(yaml_def):
    valores = {}
    for campo in yaml_def["campos"]:
        nome, tipo = campo["nome"], campo["tipo"]
        rotulo = nome + (" *" if campo.get("obrigatorio") else "")
        if tipo == "numero":
            valores[nome] = st.number_input(rotulo, min_value=float(campo.get("min", 0)),
                                             max_value=float(campo.get("max", 1e9)))
        elif tipo == "data":
            valores[nome] = str(st.date_input(rotulo))
        elif tipo == "mes":
            valores[nome] = st.selectbox(rotulo, [f"{m:02d}" for m in range(1, 13)])
        else:
            valores[nome] = st.text_input(rotulo)
    return valores
print("Formulário gerado do YAML (text/number/date/select).")
```
