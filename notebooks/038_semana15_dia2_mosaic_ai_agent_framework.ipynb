## 📖 Teoria — Mosaic AI Agent Framework

O **Mosaic AI Agent Framework** é o conjunto oficial para construir, avaliar e deployar agentes:

- **Agent Bricks**: biblioteca de avaliação (RAG/agent metrics), guardrails e tracing padronizados
- **Agent Evaluation**: avaliação automatizada com golden sets
- **Deploy**: serve o agente como endpoint (serving) com o `databricks-agents` SDK


## 📖 Teoria — Agent Bricks (2026)

O **Agent Bricks** é o 'framework de frameworks': avaliação (mlflow.evaluate com databricks-agent), guardrails prontos e review de qualidade — o padrão que empresas pedem. A DAIS 2026 anunciou o **Omnigent** (orquestração de múltiplos agentes) como evolução.


### 💻 Na prática — Avaliação com Agent Bricks

Use o SDK de agentes para empacotar e avaliar.


```python
# Empacotar o agente como código (estrutura padrão)
print("""
config.yaml (agent):
  name: agente_vendas
  model:
    name: databricks-llama-3-1-70b
  tools:
    - unity_catalog_function: workspace.prata.receita_periodo
    - unity_catalog_function: workspace.prata.top_pais
  evaluation:
    golden_set: workspace.prata.golden_set_agente
""")
print("config.yaml declara modelo, tools e avaliação — o padrão Agent Bricks.")
```

```python
# Registrar o agente (MLflow) e avaliar
import mlflow
with mlflow.start_run(run_name="agente_bricks_v1"):
    # Em produção: mlflow.langchain.log_model + mlflow.evaluate
    mlflow.log_param("framework", "agent_bricks")
    mlflow.log_param("guardrails", "offensive, pii")
    print("Agente registrado para avaliação (Agent Bricks).")
```

### 💻 Na prática — Deploy (trial)

No trial: **Agent Evaluation → Review → Deploy** — o agente vira endpoint servido com tracing e guardrails habilitados.


```python
# Deploy do agente (🔑 trial)
print("""
1. Agent Evaluation: rode o golden set (nota do agente)
2. Review: veja falhas por categoria (alucinação, tool errada)
3. Deploy: cria serving endpoint do agente (REST)
4. O endpoint aceita mensagens e retorna resposta + traces
""")
print("Na Free, avalie com mlflow.evaluate; o deploy gerenciado é 🔑.")
```
