## 📖 Teoria — pgvector no Lakebase

O Lakebase suporta **pgvector**: coluna `vector` + busca por similaridade com índices:

- **HNSW** (Hierarchical Navigable Small World): mais preciso, mais memória; recomendado para precisão
- **IVFFlat**: mais rápido de buildar, menos preciso em dados novos

Para RAG de produção, **HNSW** é o padrão.


### 💻 Na prática — Tabela com embeddings

Crie a tabela vetorial dos produtos.


```python
# Tabela vetorial com pgvector
print("""
-- No Lakebase:
CREATE EXTENSION IF NOT EXISTS vector;
CREATE TABLE produtos_vetores (
    produto_id TEXT PRIMARY KEY,
    texto TEXT,
    embedding vector(1024)
);
-- Índice HNSW (precisão)
CREATE INDEX ON produtos_vetores USING hnsw (embedding vector_cosine_ops);
""")
print("Tabela + índice HNSW prontos.")
```

```python
# Inserir embeddings
print("""
INSERT INTO produtos_vetores (produto_id, texto, embedding)
VALUES
  ('85123A', 'Copo de vidro', '[0.1, 0.2, ...]'::vector),
  ('71053', 'Panela antiaderente', '[0.4, 0.1, ...]'::vector);
""")
print("Embeddings inseridos (gerados com a FMA de embeddings).")
```

### 💻 Na prática — Busca por similaridade

Consulte os vizinhos mais próximos.


```python
# Busca vetorial (k-NN)
print("""
SELECT produto_id, texto, 1 - (embedding <=> '[0.12, 0.19, ...]') AS similaridade
FROM produtos_vetores
ORDER BY embedding <=> '[0.12, 0.19, ...]'
LIMIT 5;
""")
print("A busca retorna os itens mais similares ao embedding da pergunta.")
```

> 🎯 **Dica de prova**: pgvector: HNSW (precisão) vs IVFFlat (velocidade de build). Pergunta: 'qual índice para busca vetorial precisa?' → HNSW.
