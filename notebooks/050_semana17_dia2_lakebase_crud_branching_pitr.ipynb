## 📖 Teoria — Operações transacionais no Lakebase

O Lakebase aceita **CRUD** (insert/update/delete) com latência de app — via API (SQL/Postgres) ou SDK. É onde vive o **estado do agente** (sessões, memória, pedidos).


## 📖 Teoria — Instant Branching e PITR

- **Instant Branching**: cria branches da base em segundos (dev/teste sem duplicar dados) — como Git para bancos
- **PITR (Point-in-Time Recovery)**: voltar a base para um momento anterior — proteção contra erros


### 💻 Na prática — CRUD via SQL

Crie a tabela transacional do agente (sessões).


```python
# Estrutura da base transacional (via connection do Lakebase)
print("""
-- No console do Lakebase (ou connection):
CREATE TABLE sessoes (
    sessao_id UUID PRIMARY KEY,
    usuario STRING,
    estado STRING,      -- json do estado da conversa
    atualizado_em TIMESTAMP DEFAULT now()
);
INSERT INTO sessoes (sessao_id, usuario, estado) VALUES
    (gen_random_uuid(), 'ana', '{"turnos": 3}');
UPDATE sessoes SET estado = '{"turnos": 4}' WHERE usuario = 'ana';
SELECT * FROM sessoes;
""")
print("CRUD transacional do agente demonstrado.")
```

### 💻 Na prática — Branch e PITR (UI)

No console do Lakebase: **Branch → Create branch** (dev) e **PITR → Restore**.


```python
# Branch e PITR via API (visão geral)
print("""
POST /api/2.0/lakebase/projects/{id}/branches  {"name": "dev-teste"}
POST /api/2.0/lakebase/projects/{id}/pitr       {"restore_to": "2026-08-01T10:00:00Z"}
""")
print("Branch instantânea para testes; PITR para recuperação.")
```

> 🎯 **Dica de prova**: Lakebase: CRUD transacional + Instant Branching (dev/teste) + PITR (recuperação). Pergunta: 'como testar mudanças sem afetar prod?' → branch.
