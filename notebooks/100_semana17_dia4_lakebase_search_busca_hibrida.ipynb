## 📖 Teoria — Lakebase Search

O **Lakebase Search** junta busca lexical (BM25) e semântica (vetores) com **RRF** nativamente — sem montar a fusão manualmente (como fizemos na Semana 12). Suporta filtros de metadata.


### 💻 Na prática — Habilitando a busca

Crie o índice de texto e a busca híbrida.


```python
# Índice de texto (BM25)
print("""
-- No Lakebase:
CREATE INDEX ON produtos_vetores USING bm25 (texto);
""")
print("Índice lexical criado.")
```

```python
# Busca híbrida com RRF (conceito — a API do Lakebase Search)
print("""
POST /api/2.0/lakebase/search
{
  "query": "copo de vidro",
  "tables": ["produtos_vetores"],
  "hybrid_search": true,        // BM25 + semântica
  "ranking": "rrf",              // fusão
  "filters": {"categoria": "cozinha"}  // metadata
}
""")
print("Busca híbrida nativa com filtros.")
```

### 💻 Na prática — Comparando com a Semana 12

Na Semana 12, a fusão RRF era manual (código). No Lakebase Search, o Databricks faz nativamente — mesmo conceito, menos código e mais escala.


```python
# Mesma fusão, agora nativa
print("""
Semana 12 (manual):  rrf(busca_semantica, busca_bm25)
Lakebase Search:    hybrid_search=true + ranking=rrf
""")
print("Aprendizado anterior não foi em vão — você entende o que roda por trás.")
```

> 🎯 **Dica de prova**: Lakebase Search: híbrido (BM25+vetores) + RRF nativo + filtros. Pergunta: 'como fazer busca híbrida sem código?' → Lakebase Search.
