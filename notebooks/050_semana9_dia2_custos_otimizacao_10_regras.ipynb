## 📖 Teoria — Como o Databricks cobra

A cobrança é por **DBU (Databricks Unit)**: uma unidade de processamento conforme o tipo de workload. Fatores da fatura:
- Tipo de compute (serverless, SQL warehouse, jobs)
- Tempo de execução × tamanho do cluster
- Model serving / FMA (tokens)

**Tags** marcam recursos por projeto/time/custo. **Budget alerts** avisam quando o gasto passa do limite.


## 📖 Teoria — As 10 regras de economia

1. **Pare o que não usa** — auto-stop de warehouses/clusters (maior economia)
2. **Escolha o compute certo** — serverless para ad hoc; jobs para carga
3. **Reduza o shuffle** — broadcast join, chaves boas
4. **Use Delta/Liquid Clustering** — menos leitura, menos DBU
5. **Filtre cedo** — não leia colunas/linhas que não precisa
6. **Modo triggered** em vez de continuous quando possível
7. **Cache e result caching** para dashboards
8. **Evite UDFs** — expressões nativas
9. **Frequência certa de jobs** — diário em vez de horário
10. **Monitore** — system tables de billing, budget alerts
