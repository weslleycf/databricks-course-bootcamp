## 📖 Teoria — Full-stack nos Apps

Os Databricks Apps suportam **qualquer stack**: além de Streamlit, você pode rodar **Next.js/React** (frontend) + **FastAPI** (backend) — o padrão moderno de produto de dados.

Estrutura:
```
app/
 ├── app.yaml          # comando: next start (ou uvicorn)
 ├── frontend/         # Next.js (React)
 ├── backend/          # FastAPI
 └── requirements.txt / package.json
```


### 💻 Na prática — App Next.js mínimo

Estrutura de um app React que chama a API.


````python
# frontend/app/page.tsx (conceito Next.js)
tsx = """
'use client';
import { useEffect, useState } from 'react';

export default function Home() {
  const [resposta, setResposta] = useState("");
  useEffect(() => {
    fetch("/api/health")
      .then(r => r.json())
      .then(d => setResposta(d.status));
  }, []);
  return <h1>App de Vendas — API: {resposta}</h1>;
}
"""
print(tsx)
print("Componente React que consulta o backend.")
````

````python
# app.yaml (Next.js)
yaml = """
command:
  - next
  - start
  - -p
  - "8080"
"""
print(yaml)
````

### 💻 Na prática — Publicando

1. Suba a pasta com package.json + app.yaml.
2. **Apps → Create App**.
3. O build (npm install + next build) roda na plataforma.
