## 📖 Teoria — A arquitetura de 3 camadas

```
Next.js (UI) → FastAPI (backend) → Lakehouse (dados) / Agente
```

O frontend nunca fala com o Spark diretamente — passa pelo backend, que centraliza segurança e regras.


### 💻 Na prática — Backend que consulta o Ouro

API com endpoints de dados.


````python
# backend/main.py — endpoints de dados
from fastapi import FastAPI
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
app = FastAPI()

@app.get("/api/kpis")
def kpis():
    df = spark.table("workspace.ouro.vendas_por_dia").toPandas()
    return {"receita_total": float(df["receita_total"].sum()),
            "dias": len(df)}

@app.get("/api/top_produtos")
def top_produtos():
    return spark.table("workspace.ouro.top_produtos").limit(10).toPandas().to_dict("records")
print("Backend com /api/kpis e /api/top_produtos.")
````

````python
# Frontend consome os endpoints
tsx = """
const [kpis, setKpis] = useState(null);
useEffect(() => {
  fetch("/api/kpis").then(r => r.json()).then(setKpis);
}, []);
return <div>{kpis && `Receita total: ${kpis.receita_total}`}</div>;
"""
print(tsx)
````

### 💻 Na prática — Proxy e rotas

No app.yaml, o frontend Next.js usa rotas de API (rewrite) para o backend — ou o backend roda na mesma porta via proxy.
