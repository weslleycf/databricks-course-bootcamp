## 📖 Teoria — O que é um LLM

Um **LLM (Large Language Model)** prevê o próximo token de uma sequência. Conceitos-base:

- **Token**: unidade de texto (palavra/pedaço). 1 token ≈ 0,75 palavra em inglês.
- **Janela de contexto**: nº máximo de tokens que o modelo 'enxerga' por chamada (ex.: 8k, 128k). Passou disso = truncamento.
- **Temperature**: aleatoriedade da saída (0 = determinístico; 1 = criativo).
- **System prompt**: instruções permanentes; **user prompt**: pedido da vez.

> 🎯 **Dica de prova (GenAI Assoc)**: 'o que é a janela de contexto?' → limite de tokens de entrada+saída por chamada. Custo ≈ tokens processados.


## 📖 Teoria — Foundation Model APIs (FMA)

As **Foundation Model APIs** do Databricks dão acesso a modelos (Llama, Mistral, GPT via gateway) com **um endpoint único** e **Unity Catalog governance**: `serving_endpoint` apontando para `databricks-llama-3-1-70b`, por exemplo.

Na Free Edition, as FMAs existem **com quota de uso** (limite diário) — suficiente para o curso.


### 💻 Na prática — Primeira chamada via FMA

Use o MLflow Deployments para chamar um endpoint de FMA.


```python
# Listar endpoints de FMA disponíveis
from mlflow.deployments import get_deploy_client
client = get_deploy_client("databricks")
endpoints = client.list_endpoints()
print([e.name for e in endpoints])
```

```python
# Chamada a um LLM via FMA (ajuste o nome ao endpoint disponível)
endpoint = "databricks-llama-3-1-70b"
resp = client.predict(
    endpoint=endpoint,
    inputs={"messages": [
        {"role": "system", "content": "Você é um assistente de dados."},
        {"role": "user", "content": "Explique em 1 frase o que é o Lakehouse."}
    ], "temperature": 0.2})
print(resp["choices"][0]["message"]["content"])
```

```python
# Playground (exploratório)
print("""
1. AI > Playground
2. Escolha o modelo (ex.: Llama 3.1 70B)
3. Teste: temperature 0 vs 1, system prompts diferentes
4. Veja a contagem de tokens
""")
print("O Playground é o melhor laboratório para entender temperature.")
```
