### 💻 Na prática — MCP

Conecte o agente a ferramentas MCP (ex.: calendário, CRM) — o padrão 2026.


```python
# MCP (conceito — servidor expõe ferramentas)
print("""
1. Um servidor MCP expõe ferramentas (JSON-RPC):
   - tools/list, tools/call
2. O agente Databricks adiciona o servidor MCP como fonte de tools
3. Ex.: MCP do CRM -> o agente consulta clientes reais
""")
print("MCP = ferramentas padronizadas, interoperáveis entre LLMs.")
```
