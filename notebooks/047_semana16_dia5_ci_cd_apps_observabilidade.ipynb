## 📖 Teoria — Deploy de apps como código

Apps podem ser deployados via **DABs** (`resources.apps`) e CI/CD — o mesmo fluxo de pipelines: PR valida, merge deploya.


### 💻 Na prática — App no DABs

Adicione o app ao bundle.


```python
# databricks.yml com app
yaml = """
bundle:
  name: vendas_apps
resources:
  apps:
    app_vendas:
      name: app-vendas
      source:
        path: ./apps/dashboard
targets:
  dev:
    mode: development
  prod:
    mode: production
"""
print(yaml)
```

```python
# Deploy do app (trial/CLI)
print("""
databricks bundle deploy -t dev
databricks apps get --name app-vendas
""")
print("O app vira recurso do bundle — versionado e replicável.")
```

### 💻 Na prática — CI/CD para apps

GitHub Actions: no PR roda `bundle validate`; no merge, `bundle deploy -t prod` (mesmo padrão da Semana 6).


```python
# Observabilidade do app
print("""
1. Logs do app: Apps > app > Logs
2. Métricas: requests, erros, latência (abas do app)
3. Alerte: >5% de erros → notificar
4. Scale-to-zero: app para sozinho após inatividade (Free: 24h)
""")
print("Monitorar apps = saber se o produto está no ar e saudável.")
```

> 🎯 **Dica de prova**: Apps/DABs: `resources.apps` no bundle + CI/CD — o deploy de apps é código. Pergunta: 'como versionar o deploy de um app?' → DABs.
