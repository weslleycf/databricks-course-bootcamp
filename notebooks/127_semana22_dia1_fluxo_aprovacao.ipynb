## 📖 Teoria — O workflow de aprovação

Submissões **nunca** vão direto para os dados: passam por aprovação (o aprovador vê o resumo e decide). Estados: `pendente` → `aprovado` | `rejeitado` (com motivo).

Tudo registrado em `workspace.app.aprovacoes` — trilha completa.


### 💻 Na prática — Aprovação

Aprovar/rejeitar com motivo.


```python
# Aprovar submissão
def aprovar(sid, aprovador, status, motivo=""):
    spark.sql(f"UPDATE workspace.app.submissoes SET status = '{status}' WHERE submissao_id = '{sid}'")
    spark.createDataFrame([(sid, aprovador, "now", status, motivo)],
        ["submissao_id", "aprovado_por", "aprovado_em", "status", "motivo"])\
        .withColumn("aprovado_em", current_timestamp())\
        .write.mode("append").saveAsTable("workspace.app.aprovacoes")
    print(f"Submissão {sid[:8]} -> {status}")
aprovar("abc-123", "ana", "aprovado")
aprovar("abc-456", "ana", "rejeitado", "meta fora do padrão")
```

```python
# Pendências do aprovador
def pendentes():
    return spark.sql("SELECT submissao_id, fluxo_id, criado_por, criado_em FROM workspace.app.submissoes WHERE status = 'pendente'").toPandas()
print(pendentes())
```
