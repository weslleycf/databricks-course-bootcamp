## 📖 Teoria — O problema: dados degradam

Dados mudam com o tempo: schema muda (coluna nova), distribuição muda (drift), qualidade cai (nulos sobem). Quem consome o Ouro precisa saber ANTES de quebrar.

**Lakehouse Monitoring** cria **monitores** em tabelas que calculam métricas e **alertam em drift**:
- drift de schema (colunas novas/removidas)
- drift de dados (distribuição de valores/estatísticas)
- qualidade (expectations/constraints violadas)


## 📖 Teoria — Como funciona

1. Você cria um monitor numa tabela (`CREATE MONITOR ...` via UI).
2. O Databricks calcula perfis (contagem, min/max, quantis, nulos, tipos).
3. Execuções seguintes comparam com a **baseline** e sinalizam drift.
4. Alertas notificam a equipe.

> Na Free Edition o monitor básico existe; o avançado (agendamento, integração com alertas) roda no trial pago.


### 💻 Na prática — Criando um monitor

Pela UI: **Catalog → workspace.ouro.vendas_por_dia → Quality → Create Monitor**.


```python
# Monitor via SQL (existe em contas com o recurso)
monitor_sql = """
CREATE MONITOR workspace.ouro.vendas_por_dia
ON TABLE workspace.ouro.vendas_por_dia
WITH (metric_type = 'data_quality')
"""
print(monitor_sql)
print("Na Free, prefira a UI: Catalog > tabela > Quality > Create Monitor.")
```

```python
# Ver os perfis calculados
display(spark.sql("SELECT * FROM workspace.ouro.vendas_por_dia LIMIT 5"))
print("O monitor grava métricas em tabela system/table de perfil.")
```

### 💻 Na prática — Drift na prática

Simule um drift (mudança de distribuição) e observe.


```python
# Simular: adicionar muitas linhas de um país novo
novo_pais = spark.table("workspace.bronze.vendas_bronze").filter("Country = 'France'")
novo_pais.write.mode("append").saveAsTable("workspace.ouro.vendas_por_dia")
print("Drift simulado: distribuição por país mudou.")
```

> 🎯 **Dica de prova**: DEP: Lakehouse Monitoring cobre **drift de schema e de dados** e **quality monitors**. Pergunta: 'como detectar que a distribuição de uma coluna mudou?' → monitor de drift.
