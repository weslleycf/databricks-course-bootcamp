## 📖 Teoria — O problema das contas erradas

Assistentes de dados erram quando **inventam a fórmula** (ARPU = receita/clientes? receita/notas?). A **Genie Ontology** (DAIS 2026) resolve: você define as métricas e cálculos **uma vez** — o Genie/agente usam a definição oficial, sempre.

Na prática: uma camada semântica com métricas, dimensões e regras de negócio — o assistente consulta a ontologia antes de calcular.


### 💻 Na prática — Ontologia conceitual

Modele as métricas oficiais do projeto (mesmo conceito da ontologia).


```python
# Definição canônica das métricas (a "ontologia" do projeto)
ontologia = {
    "receita_total": "SUM(Quantity * UnitPrice) filtrado por Quantity > 0",
    "ticket_medio": "receita_total / COUNT(DISTINCT InvoiceNo)",
    "arpu": "receita_total / COUNT(DISTINCT CustomerID)",
    "n_vendas": "COUNT(*)",
}
print("Métricas canônicas definidas — todo agente usa estas fórmulas.")
```

```python
# Prompt do agente com a ontologia embutida
ontologia_prompt = """
Métricas OFICIAIS (use SEMPRE estas definições):
- receita_total = SUM(Quantity * UnitPrice) com Quantity > 0
- ticket_medio = receita_total / COUNT(DISTINCT InvoiceNo)
- arpu = receita_total / COUNT(DISTINCT CustomerID)
Nunca invente outra fórmula para essas métricas.
"""
print(ontologia_prompt)
```

### 💻 Na prática — Genie com Ontologia (trial)

No trial: **AI/BI → Genie → space → Ontology** — crie as métricas na UI e o Genie passa a usá-las nas respostas.


```python
# Fluxo da ontologia no Genie (🔑)
print("""
1. Genie space > Ontology
2. Defina: receita_total, ticket_medio, arpu (nome + SQL)
3. Associe dimensões (país, mês)
4. Pergunte no Genie: "qual o ARPU?" -> usa a definição oficial
5. O mesmo vale para o agente (via Genie Agents)
""")
print("Ontologia = a fonte única de verdade das métricas.")
```

> 🎯 **Dica de prova**: DAA/GenAI 2026: Ontologia é o antídoto para contas erradas — o assistente usa a definição canônica. Pergunta: 'como garantir que o Genie não erre cálculo?' → Ontologia.
