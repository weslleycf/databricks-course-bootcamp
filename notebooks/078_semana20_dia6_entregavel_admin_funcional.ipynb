```python
# app_admin.py (Streamlit) — esqueleto completo
import streamlit as st
st.set_page_config(page_title="DataFlow Admin", layout="wide")
st.title("⚙️ DataFlow Admin")

usuario = "admin"  # na prática: auth do workspace
st.sidebar.write(f"Usuário: {usuario}")

aba = st.sidebar.radio("Seção", ["Fluxos", "Editor YAML", "Versões"])

if aba == "Fluxos":
    st.dataframe(spark.table("workspace.app.fluxos").toPandas())
elif aba == "Editor YAML":
    fluxo = st.selectbox("Fluxo", ["metas", "descontos", "precos", "clientes"])
    yaml_atual = st.text_area("YAML", value=metas_yaml, height=300)
    if st.button("Validar e salvar"):
        try:
            validate(instance=yaml.safe_load(yaml_atual), schema=json_schema)
            salvar_versao(fluxo, yaml_atual, usuario)
            st.success("Salvo!")
        except Exception as e:
            st.error(f"Inválido: {e.message}")
elif aba == "Versões":
    st.dataframe(spark.table("workspace.app.fluxo_versoes").toPandas())
print("App admin integrado (edite, valide, salve, veja versões).")
```
