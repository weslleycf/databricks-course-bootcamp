## 📖 Teoria — Rastreabilidade e LGPD no app

- **Rastreabilidade**: cada linha tem `submissao_id`, `criado_por`, `aprovado_por` — auditável de ponta a ponta
- **LGPD**: direito ao esquecimento (apagar dados de um usuário), retenção limitada (TTL), e consentimento registrado


### 💻 Na prática — Rastreabilidade

Consulte a trilha completa de uma submissão.


```python
# Trilha completa de uma submissão
def trilha(sid):
    return spark.sql(f"""
    SELECT s.submissao_id, s.fluxo_id, s.criado_por, s.criado_em,
           s.status, a.aprovado_por, a.aprovado_em, a.motivo
    FROM workspace.app.submissoes s
    LEFT JOIN workspace.app.aprovacoes a ON s.submissao_id = a.submissao_id
    WHERE s.submissao_id = '{sid}'
    """).toPandas()
print(trilha("abc-123"))
```

### 💻 Na prática — LGPD: direito ao esquecimento

Apagar dados de um usuário (lógico + físico).


```python
# Direito ao esquecimento (por usuário)
def esquecimento(usuario):
    # 1) apagar dados pessoais do app
    spark.sql(f"DELETE FROM workspace.app.submissoes WHERE criado_por = '{usuario}'")
    # 2) apagar da tabela de clientes (se aplicável)
    # 3) logar a solicitação (auditoria de compliance)
    print(f"Esquecimento de {usuario} processado e logado.")
print("Esquecimento: DELETE + VACUUM (físico) + log da solicitação.")
```

### 💻 Na prática — Retenção e TTL

Política de retenção do app (documento) + TTL nos dados sensíveis.


```python
# Política de retenção
politica = """
- Submissões: reter 5 anos (fiscal) com TTL
- Logs de auditoria: 7 anos (compliance)
- Dados pessoais: mínimo necessário, pseudonimizados na Prata
- Esquecimento: DELETE + VACUUM + evidência logada
"""
print(politica)
```

> 🎯 **Dica de prova**: Portfólio/entrevista: LGPD no app = minimização (pseudônimo), direito ao esquecimento (delete+vacuum+log) e retenção (TTL) — as 3 respostas padrão.
