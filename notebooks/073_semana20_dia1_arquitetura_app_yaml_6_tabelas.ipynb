## 📖 Teoria — O app DataFlow Admin

O **DataFlow Admin** permite que usuários de negócio enviem dados (CSV ou formulário) com **validações 100% configuráveis via YAML por fluxo**, e uma seção admin para gerenciar os fluxos.

Arquitetura:
```
Admin (editor YAML) → Fluxos (YAML) → Motor de validação
   → aprovação → Bronze → DLT → Prata/Ouro
```


## 📖 Teoria — O modelo de 6 tabelas (Lakebase)

| Tabela | Papel |
|---|---|
| `fluxos` | definição dos fluxos (nome, campos, YAML) |
| `campos` | campos de cada fluxo (nome, tipo, regras) |
| `submissoes` | envios de dados (CSV/formulário) |
| `itens_submissao` | linhas de cada submissão |
| `validacoes` | resultados de validação por item |
| `aprovacoes` | workflow de aprovação |


### 💻 Na prática — Criando as tabelas

Crie as 6 tabelas no Lakebase (ou Delta, na Free).


````python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.app;
CREATE TABLE IF NOT EXISTS workspace.app.fluxos (
  fluxo_id STRING, nome STRING, yaml_def STRING, ativo BOOLEAN, criado_em TIMESTAMP
) USING DELTA;
CREATE TABLE IF NOT EXISTS workspace.app.campos (
  fluxo_id STRING, campo STRING, tipo STRING, obrigatorio BOOLEAN
) USING DELTA;
CREATE TABLE IF NOT EXISTS workspace.app.submissoes (
  submissao_id STRING, fluxo_id STRING, origem STRING,
  status STRING, criado_em TIMESTAMP, criado_por STRING
) USING DELTA;
CREATE TABLE IF NOT EXISTS workspace.app.itens_submissao (
  submissao_id STRING, linha INT, dados STRING
) USING DELTA;
CREATE TABLE IF NOT EXISTS workspace.app.validacoes (
  submissao_id STRING, linha INT, campo STRING, regra STRING, ok BOOLEAN, msg STRING
) USING DELTA;
CREATE TABLE IF NOT EXISTS workspace.app.aprovacoes (
  submissao_id STRING, aprovado_por STRING, aprovado_em TIMESTAMP, status STRING
) USING DELTA;
SHOW TABLES IN workspace.app;
````
