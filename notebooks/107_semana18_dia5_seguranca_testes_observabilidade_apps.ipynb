## 📖 Teoria — Segurança de apps em produção

- **RBAC**: o app respeita as permissões do UC (usuário vê só o que pode)
- **Rate limiting**: protege contra abuso de API
- **Secrets**: credenciais via secret scope (nunca no código)
- **Input validation**: sanitizar entradas (SQL injection no Text-to-SQL!)


### 💻 Na prática — Rate limit e validação

Proteja a API.


```python
# Rate limit simples no FastAPI
from fastapi import FastAPI, HTTPException
import time
app = FastAPI()
ultima_chamada = {}

@app.post("/api/perguntar")
def perguntar(p: dict):
    agora = time.time()
    if agora - ultima_chamada.get("user", 0) < 2:
        raise HTTPException(429, "Muitas requisições — aguarde 2s")
    ultima_chamada["user"] = agora
    if len(p.get("pergunta", "")) > 500:
        raise HTTPException(400, "Pergunta muito longa")
    return {"ok": True}
print("Rate limit (2s) + validação de tamanho implementados.")
```

```python
# Testes do app (pytest)
print("""
def test_health():
    r = client.get("/health")
    assert r.status_code == 200

def test_pergunta_longa():
    r = client.post("/api/perguntar", json={"pergunta": "x"*600})
    assert r.status_code == 400
""")
print("Testes básicos do app (rodam no CI).")
```

### 💻 Na prática — Observabilidade

Logs estruturados + métricas + alertas no app.


```python
# Logs estruturados
import logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("app")
logger.info("pergunta_recebida", extra={"usuario": "ana", "tamanho": 120})
print("Logs estruturados para o App Logs / Datadog.")
```
