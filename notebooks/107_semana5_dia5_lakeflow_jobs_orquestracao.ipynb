## 📖 Teoria — Lakeflow Jobs (2026)

O **Lakeflow Jobs** é o orquestrador do Databricks: roda notebooks, pipelines DLT, queries SQL e código arbitrário em sequência ou paralelo, com **dependências**, **retries**, **alertas** e **agendamento**.

Na Free Edition: **máx. 5 tarefas/jobs concorrentes** por conta.

Terminologia 2026: *Workflows* → **Lakeflow Jobs** (mesma API/UI, nome novo).


## 📖 Teoria — Componentes de um Job

- **Task**: uma unidade (notebook, pipeline, SQL query, Python wheel, ...)
- **Dependência**: task B roda após task A (via parâmetros de entrada/saída)
- **Trigger**: agendamento (cron) ou manual
- **Retry**: tentativas em falha
- **Alert**: notificação por e-mail/Slack
- **Run**: uma execução (com log, status, duração)


### 💻 Na prática — Criando o Job

1. Em **Workflows → Jobs → Create Job**.
2. Task 1: notebook `Semana 5 · Auto Loader` (ou o notebook deste curso).
3. Task 2: **Delta Live Tables pipeline** `pipeline_vendas`.
4. Task 3: query SQL que valida contagem (ex.: `SELECT COUNT(*) FROM workspace.bronze.vendas_prata`).
5. Configure dependência: Task 2 depende de Task 1; Task 3 depende de Task 2.
6. Schedule: diário às 06:00. Alert: e-mail em falha.
