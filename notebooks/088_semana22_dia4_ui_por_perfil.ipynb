## 📖 Teoria — Uma UI por perfil

O mesmo app mostra telas diferentes conforme o perfil:
- **Submissor**: formulários + status das suas submissões
- **Aprovador**: fila de pendentes + decisões
- **Admin**: editor YAML + versões + permissões

Perfil = derivado das permissões (Semana 20).


### 💻 Na prática — Roteamento por perfil

Determine o perfil e renderize a UI correspondente.


```python
# Perfil do usuário
def perfil(usuario):
    if usuario == "admin": return "admin"
    if pode(usuario, "metas", "aprovar"): return "aprovador"
    return "submissor"
print("perfil de ana:", perfil("ana"))
print("perfil de joao:", perfil("joao"))
```

```python
# UI por perfil (Streamlit)
import streamlit as st
usuario = "ana"
p = perfil(usuario)
if p == "admin":
    st.subheader("⚙️ Admin: fluxos, YAML, permissões")
    st.write("Editor de YAML + versões + RBAC")
elif p == "aprovador":
    st.subheader("✅ Aprovador: fila de pendentes")
    st.write("Lista + aprovar/rejeitar")
else:
    st.subheader("📥 Submissor: formulários")
    st.write("CSV/formulário + status")
print("UI roteada por perfil.")
```

### 💻 Na prática — Consistência

A API também valida o perfil (nunca confiar só na UI).


```python
# Backend valida o perfil
def api_aprovar(usuario, sid):
    if not pode(usuario, "metas", "aprovar"):
        return {"erro": "403 — sem permissão"}
    return aprovar(sid, usuario, "aprovado")
print("Backend também bloqueia (defesa em profundidade).")
```
