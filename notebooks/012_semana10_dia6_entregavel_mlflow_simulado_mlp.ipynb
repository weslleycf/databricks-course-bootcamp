````python
# 1) Treinar e registrar (autolog)
import mlflow
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
mlflow.autolog()
X = df[["dia_semana", "dia_mes", "mes", "ano"]]
y = df["receita_total"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
with mlflow.start_run(run_name="entrega_final"):
    m = RandomForestRegressor(n_estimators=200, max_depth=10, random_state=42)
    m.fit(X_train, y_train)
    mlflow.sklearn.log_model(m, "model")
print("Modelo treinado.")
````

````python
# 2) Registrar no UC com alias champion
from mlflow.tracking import MlflowClient
client = MlflowClient()
uri = "runs:/" + mlflow.last_active_run().info.run_id + "/model"
v = mlflow.register_model(uri, "workspace.prata.modelo_previsao_receita")
client.set_registered_model_alias("workspace.prata.modelo_previsao_receita", "champion", str(v.version))
print(f"Versão {v.version} promovida a champion.")
````

````python
# 3) Avaliar
with mlflow.start_run(run_name="avaliacao_final"):
    mlflow.evaluate(
        model="models:/workspace.prata.modelo_previsao_receita@champion",
        data=X_test.assign(receita_total=y_test),
        targets="receita_total",
        model_type="regressor")
print("Avaliação final registrada.")
````
