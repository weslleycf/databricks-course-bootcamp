## 📖 Teoria — Model Registry e UC models

O **Model Registry** (2026: **MLflow Model Registry in Unity Catalog**) é o catálogo de modelos governados:

- **UC models**: modelos registrados como objetos do UC (catalog.schema.model) — com permissões, linhagem e versionamento
- **Aliases/stages**: `alias = 'champion'` (produção) vs `'challenger'` (candidato)
- **Versionamento**: cada registro vira `Version 1, 2, 3...`

> 🎯 **Dica de prova (MLP 2026)**: a migração do registry de workspace para UC é tema recorrente. Memorize: modelo no UC = `catalog.schema.model`.


### 💻 Na prática — Registrando o modelo

Registre o modelo treinado como UC model com alias.


```python
# Registrar o modelo treinado (continue da célula anterior)
import mlflow
# O autolog já registrou o modelo; promova para o UC:
modelo_uri = "runs:/" + mlflow.last_active_run().info.run_id + "/model"
mlflow.register_model(model_uri, "workspace.prata.modelo_previsao_receita")
print("Modelo registrado no UC: workspace.prata.modelo_previsao_receita")
```

```python
# Definir alias (champion = produção)
from mlflow.tracking import MlflowClient
client = MlflowClient()
client.set_registered_model_alias("workspace.prata.modelo_previsao_receita", "champion", "1")
print("Alias champion apontando para a versão 1.")
```
