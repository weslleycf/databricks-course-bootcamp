## 📖 Teoria — UX de chat de agente

Um chat de agente profissional tem:
- **Streaming**: a resposta aparece token a token (percepção de velocidade)
- **Estados**: 'pensando', 'consultando dados', 'respondendo'
- **Ferramentas visíveis**: 'usou a tool X' (transparência)
- **Feedback**: 👍/👎 gravado

```python
# Frontend: streaming + estados
tsx = """
const [resposta, setResposta] = useState("");
const [estado, setEstado] = useState("idle"); // idle|pensando|respondendo

async function perguntar(texto) {
  setEstado("pensando"); setResposta("");
  const r = await fetch("/api/agente/stream", {method: "POST",
    body: JSON.stringify({pergunta: texto})});
  const reader = r.body.getReader();
  setEstado("respondendo");
  while (true) {
    const {done, value} = await reader.read();
    if (done) break;
    setResposta(prev => prev + new TextDecoder().decode(value));
  }
  setEstado("idle");
}
"""
print(tsx)
```
