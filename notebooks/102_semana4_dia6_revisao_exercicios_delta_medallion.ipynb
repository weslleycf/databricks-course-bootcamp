````python
# 3. MERGE de atualização na dim_cliente (SCD1)
fonte = spark.createDataFrame([("12345", "CAMPINAS")], ["CustomerID", "cidade_nova"])
fonte.createOrReplaceTempView("f")
spark.sql("""
MERGE INTO workspace.prata.dim_cliente t USING f ON t.CustomerID = f.CustomerID
WHEN MATCHED THEN UPDATE SET t.Country = f.cidade_nova
""")
print("MERGE executado (SCD1 simples).")
````
