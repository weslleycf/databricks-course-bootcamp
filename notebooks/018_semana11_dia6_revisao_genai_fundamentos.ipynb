## 🗝️ Gabarito comentado

**1.** RAG em 2 min

Recupera trechos relevantes dos seus dados (vetores) e dá ao LLM como contexto para responder com fonte — sem fine-tuning.

**2.** Erros e mitigação

1) chunk irrelevante → melhor chunking/rerank; 2) contexto fora da janela → chunk menor; 3) prompt fraco → few-shot/restrições.
