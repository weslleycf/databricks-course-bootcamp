## 📖 Teoria — Por que o deploy exige a conta paga

O `bundle deploy` cria recursos reais (jobs, pipelines) no workspace via API. A Free Edition limita jobs concorrentes e não expõe a mesma API de deploy — por isso, **este dia usa o trial pago** (14 dias) ou uma conta corporativa.

> ⚠️ **Estratégia de trial**: concentre seus dias de trial nas Semanas 6, 7, 9, 13, 15 e 19 — assim os 14 dias rendem o máximo.

## 📖 Teoria — Boas práticas de produção

- `mode: development` (dev) permite overwrite de recursos; `mode: production` protege (exige confirmação).
- Use **variáveis** (`${var.nome}`) para diferenças por ambiente (ex.: catálogo `workspace_dev` vs `workspace_prod`).
- Nunca rode `bundle destroy` sem revisar — remove recursos.


> 🎯 **Dica de prova**: Pergunta DEP típica: 'como implantar o mesmo job em 3 ambientes?' → targets + variáveis no databricks.yml + bundle deploy por target.
