## 📖 Teoria — BI em app vs dashboard SQL

O **dashboard do Databricks SQL** é rápido e sem código. O **app Streamlit** dá liberdade total (filtros custom, layout, interação). Use os dois: SQL para consumo rápido, app para produto.


### 💻 Na prática — App de BI completo

KPIs + gráficos + filtro.


```python
# app.py — BI completo
import streamlit as st
from pyspark.sql import SparkSession
import pandas as pd
spark = SparkSession.builder.getOrCreate()

st.set_page_config(page_title="BI Vendas", layout="wide")
st.title("📊 BI Vendas")

@st.cache_data
def carregar():
    return spark.table("workspace.ouro.vendas_por_dia").toPandas()

df = carregar()
receita_total = df["receita_total"].sum()
c1, c2, c3 = st.columns(3)
c1.metric("Receita total", f"R$ {receita_total:,.0f}")
c2.metric("Dias com dados", len(df))
c3.metric("Ticket médio", f"R$ {df['n_notas'].mean():,.0f}")

st.line_chart(df.set_index("data_venda")["receita_total"])
st.bar_chart(df.set_index("data_venda")["n_vendas"])
print("App de BI com KPIs e gráficos pronto.")
```

### 💻 Na prática — Filtro de período

Adicione seleção de mês.


```python
# Filtro de mês no app
import streamlit as st
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
df = spark.table("workspace.ouro.vendas_por_dia").toPandas()
df["mes"] = pd.to_datetime(df["data_venda"]).dt.month
mes = st.selectbox("Mês", sorted(df["mes"].unique()))
st.write(df[df["mes"] == mes].tail(10))
print("Filtro de mês adicionado.")
```
