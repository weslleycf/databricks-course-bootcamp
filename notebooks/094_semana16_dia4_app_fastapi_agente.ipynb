## 📖 Teoria — Arquitetura app + API

Em produção, separa-se UI (Streamlit/React) de **backend (API)** — o app chama a API, que chama o agente/lakehouse. Isso permite: reuso da API, escala independente e segurança centralizada.

O **FastAPI** é o padrão Python para isso (async, OpenAPI automático).


### 💻 Na prática — API FastAPI do agente

Crie o backend que chama o agente de vendas.


```python
# main.py (FastAPI) — API do agente
from fastapi import FastAPI
from pydantic import BaseModel
from langchain_community.chat_models import ChatDatabricks

app = FastAPI(title="API de Vendas")

class Pergunta(BaseModel):
    texto: str

@app.get("/health")
def health():
    return {"status": "ok"}

@app.post("/perguntar")
def perguntar(p: Pergunta):
    llm = ChatDatabricks(endpoint="databricks-llama-3-1-70b")
    return {"resposta": llm.invoke(p.texto).content}
print("API FastAPI com /health e /perguntar.")
```

```python
# requirements.txt do app
print("""
fastapi
uvicorn
pyspark
langchain
langchain-community
databricks-vector-search
""")
```

### 💻 Na prática — Publicando

1. Pasta do app com `main.py` + `app.yaml` (comando: `uvicorn main:app --host 0.0.0.0 --port 8080`).
2. **Apps → Create App**.
3. Teste: `https://<app>.cloud.databricks.com/health`.


```python
# app.yaml para FastAPI
yaml = """
command:
  - uvicorn
  - main:app
  - --host
  - 0.0.0.0
  - --port
  - "8080"
"""
print(yaml)
```
