## 📖 Teoria — Documentos empresariais no RAG

PDFs, planilhas e Docs são o caso mais comum de RAG. O fluxo:

1. **Parse**: extrair texto/tabelas (pypdf, python-docx, openpyxl)
2. **Chunking**: dividir preservando estrutura (por seção/tabela)
3. **Metadata**: título, página, data — para filtro e citação
4. **Embed + indexar**: igual ao pipeline de texto


## 📖 Teoria — Chunking hierárquico e semântico

- **Hierárquico**: chunks pequenos com referência ao documento/contexto pai — melhor para documentos longos
- **Semântico**: junta frases até quebrar por similaridade — chunks 'coesos' por assunto


### 💻 Na prática — Parse de PDF (exemplo)

Extraia texto de um PDF e crie chunks com metadata.


```python
# Parse de PDF (instale pypdf se necessário)
try:
    from pypdf import PdfReader
    leitor = PdfReader("/Volumes/workspace/prata/vol_modelos/exemplo.pdf")
    paginas = [p.extract_text() for p in leitor.pages]
    print("Páginas extraídas:", len(paginas))
except Exception as e:
    print("Coloque um PDF no volume primeiro. Dica:", str(e)[:80])
```

```python
# Chunking por página + metadata
chunks = []
for i, txt in enumerate(paginas[:10] if "paginas" in dir() else []):
    if txt:
        chunks.append({"id": i, "pagina": i+1, "texto": txt[:800],
                       "fonte": "exemplo.pdf"})
print("Chunks:", len(chunks))
for c in chunks[:2]:
    print(c["pagina"], "|", c["texto"][:80])
```
