## 📖 Teoria — MCP na prática

O **MCP** (Model Context Protocol, aberto em 2024) padroniza como LLMs acessam ferramentas e dados: um servidor MCP expõe **tools** via JSON-RPC, e qualquer cliente MCP (incluindo agentes Databricks) as consome.

Em 2026 é o padrão de integração: CRM, calendário, bancos, APIs internas — tudo vira tool MCP.


### 💻 Na prática — Servidor MCP mínimo

Estrutura de um servidor MCP que expõe tools.


```python
# servidor_mcp.py (conceito — usa a lib mcp)
print("""
from mcp.server.fastmcp import FastMCP
mcp = FastMCP("vendas")

@mcp.tool()
def receita_por_pais(pais: str) -> str:
    """Receita de um país (Ouro)."""
    ...

mcp.run()
""")
print("Servidor MCP expõe receita_por_pais como tool.")
```

```python
# Conectar o agente ao servidor MCP
print("""
1. Rode o servidor MCP (URL/SSE)
2. No Databricks: AI > Agents > Add MCP server
3. O agente lista as tools do servidor e as chama
4. Ex.: MCP do CRM -> agente consulta clientes reais
""")
print("MCP = ferramentas interoperáveis entre qualquer LLM.")
```
