## 📖 Teoria — O que são DABs

**Databricks Asset Bundles (DABs)** é o formato de **Infraestrutura como Código** (IaC) do Databricks: você define jobs, pipelines, notebooks e apps em YAML, e o Databricks **implantar** tudo com `bundle deploy`.

Vantagens: versionado no Git, replicável em N ambientes, revisável em PR, e padrão oficial de produção (DEP).


## 📖 Teoria — Estrutura de um bundle

```
meu_bundle/
 ├── databricks.yml        # definição principal (recursos, targets)
 ├── resources/
 │   ├── jobs/             # jobs do bundle
 │   └── pipelines/        # pipelines DLT
 └── src/                  # código-fonte (notebooks, .py)
```

`databricks.yml` tem 3 blocos-chave: `bundle` (nome), `resources` (o que implantar), `targets` (ambientes dev/staging/prod com variáveis).


### 💻 Na prática — Criando o bundle

Com o **Databricks CLI** instalado (`brew install databricks` / `pip install databricks-cli` — use o novo `databricks` CLI):


````python
# Terminal local — iniciar bundle
!mkdir -p ~/bundle_vendas && cd ~/bundle_vendas
!databricks bundle init default-python
print("Bundle criado: databricks.yml + src/ + resources/")
````

````python
# databricks.yml mínimo (referência — crie/edit no seu projeto)
yaml = """
bundle:
  name: vendas_bundle
resources:
  jobs:
    job_ingestao:
      name: job_ingestao
      tasks:
        - task_key: t1
          notebook_task:
            notebook_path: ./src/ingestao.ipynb
targets:
  dev:
    mode: development
  prod:
    mode: production
"""
print(yaml)
````

### 💻 Na prática — Validate e plan

Valide o bundle localmente (sem deploy) — isso roda na Free Edition.


````python
# Terminal local
!databricks bundle validate -t dev
!databricks bundle plan -t dev
print("Validate/plan checam sintaxe, recursos e variáveis sem deploy.")
````
