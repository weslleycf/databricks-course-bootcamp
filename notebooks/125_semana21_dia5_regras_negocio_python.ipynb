## 📖 Teoria — Regras declaradas no YAML

O YAML declara regras de negócio que o motor executa:

```yaml
regras:
  - campo: percentual
    op: lte
    valor: 100
  - campo: meta
    op: gte
    valor: 0
```

Assim, o negócio muda regras **sem deploy de código**.


### 💻 Na prática — Motor de regras

Implemente um mini-avaliador de regras.


````python
# Avaliador de regras do YAML
def aplicar_regras(row, regras):
    erros = []
    ops = {"eq": lambda a, b: a == b, "neq": lambda a, b: a != b,
           "gt": lambda a, b: a > b, "gte": lambda a, b: a >= b,
           "lt": lambda a, b: a < b, "lte": lambda a, b: a <= b}
    for regra in regras:
        campo = regra["campo"]
        valor = row.get(campo)
        try:
            ok = ops[regra["op"]](float(valor), float(regra["valor"]))
            if not ok:
                erros.append(f"{campo} falhou {regra['op']} {regra['valor']}")
        except (TypeError, ValueError):
            erros.append(f"{campo} incomparável")
    return erros
print("Avaliador de regras pronto.")
````

````python
# Regras do fluxo descontos (do YAML)
regras_descontos = [
    {"campo": "percentual", "op": "lte", "valor": 100},
    {"campo": "percentual", "op": "gte", "valor": 0},
]
print("regras:", aplicar_regras({"percentual": 150}, regras_descontos))
print("regras:", aplicar_regras({"percentual": 15}, regras_descontos))
````

### 💻 Na prática — Expressões livres

Suporte a expressões simples (eval seguro em sandbox).


````python
# Expressão (sandbox: sem builtins perigosos)
import ast
def avaliar_expr(expr, row):
    arvore = ast.parse(expr, mode="eval")
    nomes = {n.id: row.get(n.id) for n in ast.walk(arvore) if isinstance(n, ast.Name)}
    return eval(compile(arvore, "<expr>", "eval"), {"__builtins__": {}}, nomes)
print(avaliar_expr("percentual <= 100 and percentual >= 0", {"percentual": 15}))
print("Expressões avaliadas sem builtins (sandbox).")
````

> 🎯 **Dica de prova**: Portfólio: regras declarativas no YAML (op/valor) + expressões sandbox = negócio configura sem código. Pergunta: 'como o negócio muda regras sem deploy?' → YAML + motor.
