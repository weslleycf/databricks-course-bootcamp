## 📖 Teoria — BI direto na fonte: a vantagem do Lakehouse

No Lakehouse, os dashboards consultam **as mesmas tabelas Delta** que os pipelines escrevem — sem cópias para um BI separado. Isso elimina duplicidade, atraso e divergência de números.

O **Databricks SQL** permite: visualizações, dashboards, alertas e agendamento de queries — tudo sobre o Unity Catalog.


### 💻 Na prática — Criando o primeiro dashboard

1. Abra **SQL Warehouses** e garanta que o warehouse está **Running** (2X-Small).
2. Vá em **Queries → New Query** e rode:
```sql
SELECT Country, COUNT(*) AS vendas, ROUND(SUM(Quantity*UnitPrice),2) AS receita
FROM workspace.bronze.vendas_bronze GROUP BY Country ORDER BY receita DESC LIMIT 10;
```
3. No resultado, clique em **Visualization** e escolha **Bar chart**.
4. Clique em **Save → Create dashboard** e adicione mais 2–3 queries (vendas por mês, top produtos).
5. Publique o dashboard (botão **Publish**).

## 🗝️ Gabarito comentado

**1.** Query agendada

Em Queries: botão **Schedule** → frequência diária 08:00 → **Refresh**. Com destino e-mail configurados no workspace, o resultado chega por e-mail.

**2.** Alerta de queda de receita

Criar query com `(receita_hoje - receita_ontem)/receita_ontem`, e em **Alerts** definir condição (< -0.30) e ação (notificação por e-mail).

**3.** Bronze vs Ouro

O Ouro é modelado, limpo, denormalizado e estável para negócio; o Bronze é append-only e pode ter erros/lixo. Consultar Ouro evita respostas erradas e reprocessamentos.
