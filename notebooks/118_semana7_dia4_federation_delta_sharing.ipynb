## 📖 Teoria — Lakehouse Federation

O **Lakehouse Federation** permite consultar dados que vivem **fora** do Databricks (Postgres, Snowflake, BigQuery, MySQL...) como se fossem tabelas do UC, via **connections** — sem copiar dados.

Casos: migração gradual, dados operacionais, eliminar ETL de cópia.

> ⚠️ Federation exige conta paga (connections a fontes externas não existem na Free Edition).


## 📖 Teoria — Delta Sharing

O **Delta Sharing** é o protocolo aberto de compartilhamento de dados: você compartilha tabelas com outras organizações (Databricks ou não) via **shares** e **recipients**, com controle de acesso, sem copiar dados.

Na Free Edition, você pode **receber** shares (se o provedor te adicionar) e entender o protocolo; criar shares como provedor é recurso de conta completa.


### 💻 Na prática — Federation (trial pago)

No workspace pago:


```python
# 1) Criar connection via UI
print("""
Catalog > Add > Connection
Escolha o tipo (Postgres/MySQL/Snowflake/...)
Preencha host, porta, banco, usuário e senha (via secret scope!)
""")
print("Depois: CREATE FOREIGN CATALOG postgres FROM CONNECTION minha_conn;")
```

```python
# 2) Consultar dados federados
query = """
-- A tabela remota vira catalog.schema.table normalmente
CREATE FOREIGN CATALOG meus_dados FROM CONNECTION conn_postgres;
SELECT * FROM meus_dados.public.clientes LIMIT 10;
"""
print(query)
print("Consultas federadas usam a mesma sintaxe — o UC faz a ponte.")
```

### 💻 Na prática — Delta Sharing (conceito + receber)

Como **provedor** (paga): Catalog → Delta Sharing → Create Share → add tables → Create Recipient (nome + token) → enviar token ao parceiro.


```python
# Como RECEPTOR (Free pode receber)
print("""
1. O provedor te envia o token (share credential)
2. Catalog > Delta Sharing > Open shared data
3. Cole o token -> a tabela compartilhada aparece
4. Consulte como tabela normal (somente leitura)
""")
print("Delta Sharing usa o protocolo aberto — qualquer engine Delta lê.")
```

> 🎯 **Dica de prova**: DEP 2026: Federation (connections → foreign catalogs) e Delta Sharing (shares/recipients, protocolo aberto) são perguntas recorrentes. Memorize a diferença: Federation = LER dados externos; Sharing = COMPARTILHAR dados com outros.

## 🗝️ Gabarito comentado

**1.** Federation vs Sharing

Federation: consultar dados de fora (inbound, read). Sharing: expor seus dados a outros (outbound, com controle). São duas direções da mesma ponte.

**2.** Compartilhamento

Ex.: Ouro de vendas agregadas por país (KPIs) para um parceiro de BI — sem enviar PII do Bronze.

**3.** Sem cópia

O receptor lê os arquivos Delta diretamente do storage via token — o provedor só gerencia o acesso, sem duplicar dados.
