## 📖 Teoria — A estrutura de 6 partes de um prompt profissional

1. **Papel** (role): 'Você é um analista de dados sênior'
2. **Tarefa** (task): o que fazer, verbo claro
3. **Contexto**: dados/tabelas disponíveis
4. **Restrições**: formato, tom, o que NÃO fazer
5. **Formato de saída**: JSON, markdown, tabela
6. **Exemplo** (few-shot): 1–3 exemplos

Quanto mais específico, menos alucinação.


## 📖 Teoria — Few-shot, Chain-of-Thought e JSON

- **Few-shot**: dar exemplos no prompt ensina o padrão (ótimo p/ SQL)
- **Chain-of-Thought (CoT)**: pedir 'pense passo a passo' melhora raciocínio — mas cuidado com custo/tokens
- **JSON mode**: instruir saída JSON válida (parse automático)


### 💻 Na prática — Prompt estruturado

Monte o prompt profissional para o assistente de vendas.


````python
# Prompt estruturado em 6 partes
system_prompt = """
Você é um analista de dados sênior de uma rede de varejo.
Tarefa: responder perguntas sobre vendas usando as tabelas abaixo.
Tabelas disponíveis:
- workspace.ouro.vendas_por_dia (data_venda, receita_total, n_vendas, n_notas)
- workspace.ouro.receita_por_pais (Country, receita_total, n_vendas)
- workspace.ouro.top_produtos (sk_produto, StockCode, Description, receita_total)
Regras:
- Receita = Quantity * UnitPrice.
- Se não souber, diga que não sabe (nunca invente).
- Responda em português.
Formato: resposta curta + SQL usado (bloco de código).
Exemplo:
P: qual o país com mais receita?
R: O país com mais receita é o United Kingdom.
```sql\nSELECT Country FROM workspace.ouro.receita_por_pais ORDER BY receita_total DESC LIMIT 1;\n```
"""
print(system_prompt)
````

````python
# Testar o prompt via FMA
from mlflow.deployments import get_deploy_client
client = get_deploy_client("databricks")
resp = client.predict(
    endpoint="databricks-llama-3-1-70b",
    inputs={"messages": [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": "Qual a receita de novembro?"}
    ], "temperature": 0})
print(resp["choices"][0]["message"]["content"])
````
