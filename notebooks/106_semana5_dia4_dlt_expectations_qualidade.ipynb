## 📖 Teoria — Expectations — qualidade como código

No DLT, qualidade é **declarada** junto com a tabela, com 3 níveis de severidade:

| Nível | Comportamento | Uso |
|---|---|---|
| `@dlt.expect` | conta violações e **mantém** as linhas ruins | monitorar sem bloquear |
| `@dlt.expect_or_drop` | **descarta** as linhas violadas | limpeza silenciosa |
| `@dlt.expect_or_fail` | **falha o pipeline** se violar | regra de negócio crítica |

As violações viram **métricas de qualidade** na UI do pipeline (dashboard de expectations) — auditoria automática.


### 💻 Na prática — Expectations na prática

Adicione expectations ao pipeline da tabela Prata.


```python
# ===== workspace_file: pipeline_vendas_qualidade.py =====
import dlt
from pyspark.sql.functions import col

@dlt.table(comment="Prata com qualidade")
@dlt.expect("quantidade_positiva", "Quantity > 0")
@dlt.expect_or_drop("preco_positivo", "UnitPrice > 0")
@dlt.expect_or_fail("cliente_obrigatorio", "CustomerID IS NOT NULL")
def vendas_prata():
    return (dlt.read_stream("vendas_bronze")
        .filter(col("Quantity") > 0))

@dlt.table(comment="Métricas de qualidade")
def qualidade_vendas():
    return (dlt.read("vendas_prata")
        .groupBy("Country")
        .count())
```
