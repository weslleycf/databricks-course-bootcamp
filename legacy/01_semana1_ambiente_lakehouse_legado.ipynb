# Dia 1: O que é o Lakehouse e criação da conta

### 📖 Teoria: por que Lakehouse?**Data Warehouse** (anos 90): excelente para SQL e BI, mas caro, proprietário e péssimo paraIA/ML (não lê dados não estruturados, não escala para dados de ML).**Data Lake** (anos 2010): barato, escala, lê qualquer formato — mas vira um **pântano de dados**(swamp): sem transações, sem garantia de consistência, sem governança.**Lakehouse** (Databricks, 2020): a união dos dois. Uma única plataforma que:- Guarda os dados em **formato aberto** (Parquet/Delta) em armazenamento barato (S3/ADLS/GCS)- Traz **transações ACID**, versionamento e governança de um warehouse- Roda **SQL, Python, R, Scala e IA** no mesmo dado- Elimina a cópia dupla (ELT em vez de ETL)**As 7 características do Lakehouse** (segundo a Databricks):1. Transações ACID2. Schema enforcement e evolução3. Dados abertos e acessíveis4. Suporte a BI direto na fonte5. Armazenamento decoupled de compute (S3 + clusters)6. Versionamento de dados (Time Travel)7. Governança unificada (Unity Catalog)### 🏗️ Arquitetura do Lakehouse```┌─────────────────────────────────────────────────┐│              UNITY CATALOG (Governança)          ││  catálogos → schemas → tabelas → funções → ML   │├─────────────────────────────────────────────────┤│   Compute: SQL Warehouse | Clusters | Serverless │├─────────────────────────────────────────────────┤│   Motor: Apache Spark + Delta Lake + Photon      │├─────────────────────────────────────────────────┤│   Armazenamento: DBFS / S3 / ADLS / GCS          ││   Formatos: Parquet, Delta, JSON, CSV, Avro...   │└─────────────────────────────────────────────────┘```### 🎓 Contexto do nosso projetoVocê é engenheiro(a) de dados de uma **rede de varejo** que vende produtos em múltiplas lojas.Os dados chegam em CSV (vendas), JSON (clientes) e feeds de catálogo. # Dia 4: JSON, Parquet e Delta — os 4 formatos na prática

### 🧩 Ler e gravar nos 4 formatosVamos praticar gravando a mesma tabela em 4 formatos e comparando:

````python
# 1) Gravar em Parquetdf_vendas_br.write.mode("overwrite").parquet("/tmp/vendas.parquet")print("Parquet gravado.")# 2) Gravar em JSONdf_vendas_br.limit(100).write.mode("overwrite").json("/tmp/vendas.json")print("JSON gravado (100 linhas, demonstração).")# 3) Gravar em Delta (a estrela do Lakehouse!)df_vendas_br.write.mode("overwrite").format("delta").save("/tmp/vendas_delta")print("Delta gravado.")# 4) Comparar tamanhosdisplay(spark.sql("""    SELECT 'parquet' AS formato, COUNT(*) AS linhas FROM parquet.`/tmp/vendas.parquet`    UNION ALL    SELECT 'delta', COUNT(*) FROM delta.`/tmp/vendas_delta`"""))
````

### 💾 Tamanhos em disco (por que Delta/Parquet são melhores)No `%fs ls` abaixo compare os tamanhos — o CSV original costuma ser 2–3x maior que o Parquet:

# Dia 5: Unity Catalog + Tabelas Delta gerenciadas (Bronze)

### 🏛️ Unity Catalog (UC) — o catálogo unificado- **Catálogo** (nível mais alto): ex.: `main` (padrão da Community Edition)- **Schema**: agrupamento lógico, ex.: `bronze`, `prata`, `ouro`- **Tabela**: dados + metadados (Delta)- **View / Função / Modelo**: objetos governados```main (catálogo) └── bronze (schema)     ├── vendas_bronze     └── voos_bronze └── prata (schema) └── ouro (schema)```**Padrão de nomenclatura**: `catalog.schema.tabela` → `main.bronze.vendas_bronze`### ⚙️ Atividade 5: criar schemas e as primeiras tabelas DeltaVamos criar a estrutura do projeto:

````python
%sql-- Criar os 3 schemas da arquitetura MedallionCREATE SCHEMA IF NOT EXISTS main.bronze;CREATE SCHEMA IF NOT EXISTS main.prata;CREATE SCHEMA IF NOT EXISTS main.ouro;SHOW SCHEMAS IN main
````

Agora, a **tabela `vendas_bronze`** — a matéria-prima do nosso projeto:

````python
%sqlCREATE OR REPLACE TABLE main.bronze.vendas_bronze (  InvoiceNo      STRING,  StockCode      STRING,  Description    STRING,  Quantity       INT,  InvoiceDate    TIMESTAMP,  UnitPrice      DOUBLE,  CustomerID     STRING,  Country        STRING,  _ingested_at   TIMESTAMP DEFAULT current_timestamp())USING DELTA;
````

# Dia 6: Arquitetura Medallion (Bronze → Prata → Ouro)

### 🥉🥈🥇 O que é a arquitetura Medallion?É o padrão oficial da Databricks para organizar dados em camadas de qualidade crescente:| Camada | Conteúdo | Quem consome | Exemplo ||---|---|---|---|| **Bronze** | Dados **brutos**, como chegaram (append-only) | Engenheiros | CSV/JSON cru, com `_ingested_at` || **Prata** | Dados **limpos**, validados, modelados (dimensionais) | Cientistas/Engenheiros | `dim_cliente`, `fato_vendas` || **Ouro** | Agregados de **negócio**, prontos para consumo | Analistas/BI/IA | `vendas_por_dia`, KPIs |**Regras de ouro da Medallion**:1. Bronze = **append-only** (nunca sobrescrever o histórico cru)2. Prata = **idempotente** (rodar 2x = mesmo resultado)3. Ouro = **denormalizado**, amigável para BI (Star Schema)4. Dados transitam **somente em uma direção**: Bronze → Prata → Ouro```┌──────────┐   ELT   ┌──────────┐   joins/    ┌──────────┐│  BRONZE  │ ──────► │  PRATA   │  agregações │   OURO   ││ cru,     │         │ limpo,   │  ─────────► │ KPIs,    ││ append   │         │ modelado │             │ BI/IA    │└──────────┘         └──────────┘             └──────────┘```### 🧠 Por que isso importa para o mercado?- **Entrevistas**: Medallion é o padrão mais cobrado em vagas Databricks- **Certificação DE Associate**: cai direto na prova- **Produção**: desacopla ingestão (Bronze) de modelagem (Prata/Ouro) — times diferentes  evoluem cada camada sem quebrar a outra### 🎯 Exercício do dia 6 — mapear o projetoCom o que você aprendeu hoje, desenhe (em um papel ou markdown):1.
